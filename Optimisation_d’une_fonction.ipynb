```python
import numpy as np
import matplotlib.pyplot as plt

# =========================
# 1. Définition de la fonction
# =========================
def f(x, y):
    return x**2 + y**2 - 2*x - 4*y
```

```python


# 2. Gradient

def gradient(x, y):
    df_dx = 2*x - 2
    df_dy = 2*y - 4
    return np.array([df_dx, df_dy])


# 3. Hessienne

def hessian():
    return np.array([[2, 0],
                      [0, 2]])

# 4. Point critique

# Résolution ∇f = 0
x_star = 1
y_star = 2

print("Point critique :", (x_star, y_star))
```

```python

# 5. Nature du point critique
H = hessian()
eigenvalues = np.linalg.eigvals(H)

if np.all(eigenvalues > 0):
    nature = "Minimum local (et global)"
elif np.all(eigenvalues < 0):
    nature = "Maximum local"
else:
    nature = "Point selle"

print("Nature du point :", nature)
```

```python

# 6. Visualisation 3D
x = np.linspace(-1, 3, 100)
y = np.linspace(0, 4, 100)
X, Y = np.meshgrid(x, y)
Z = f(X, Y)

fig = plt.figure()
ax = fig.add_subplot(projection='3d')
ax.plot_surface(X, Y, Z, alpha=0.8)
ax.scatter(x_star, y_star, f(x_star, y_star), color='black', s=50)

ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_zlabel("f(x,y)")
ax.set_title("Optimisation d'une fonction")

plt.show()
```
